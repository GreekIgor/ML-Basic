## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
# ВАШ КОД ЗДЕСЬ
import re
def convertSnakeCaseToPascalCase(convertStr:str, formatConvert:str = None):
    # определяем что это снеэк касе
    is_snake = re.fullmatch(r'[a-z][a-z0-9]*(_[a-z0-9]+)*', convertStr) is not None
    # определим что это паскал касе
    is_pascal = re.fullmatch(r'[A-Z][a-zA-Z0-9]*', convertStr) is not None

    # Если указан целевой формат — используем его
    if formatConvert == 'snake':
        target = 'snake'
    elif formatConvert == 'pascal':
        target = 'pascal'
    else:
        # Автоматическое определение
        if is_snake:
            target = 'pascal'
        elif is_pascal:
            target = 'snake'

    if target == 'pascal':
        # snake_case -> PascalCase
        return ''.join(word.capitalize() for word in convertStr.split('_'))
    else:
        # PascalCase -> snake_case
        s1 = re.sub('(.)([A-Z][a-z]+)', r'\1_\2', convertStr)
        s2 = re.sub('([a-z0-9])([A-Z])', r'\1_\2', s1)
        return s2.lower()
        
a = input("Введите строку")
print(convertSnakeCaseToPascalCase(a))
    
    
```

## Задание 2: Проверка валидности даты

Написать функцию проверяющую валидность введенной даты.

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
# ВАШ КОД ЗДЕСЬ
from datetime import datetime
def isValidDate(date:str):
    try:
        datetime.strptime(date, "%d.%m.%Y")
        return True
    except ValueError:
        return False
a = input("Введите дату")
print(isValidDate(a))
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
# ВАШ КОД ЗДЕСЬ
def isSimpleNumber(num: int):
    if num < 2:
        return False
    if num == 2:
        return True
    if num % 2 == 0:
        return False
    i = 3
    while i * i <= num:
        if num % i == 0:
            return False
        i += 2
    return True

a = int(input("Введите число"))
print(isSimpleNumber(a))
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
# ВАШ КОД ЗДЕСЬ
def format_name(name: str) -> str:
    if not name.isalpha():
        raise ValueError("Имя должно содержать только буквы.")
    return name.capitalize()

def validate_age(age_str: str) -> int:
    try:
        age = int(age_str)
    except ValueError:
        raise ValueError("Возраст должен быть целым числом.")
    if not (18 <= age <= 60):
        raise ValueError("Возраст должен быть от 18 до 60.")
    return age

def format_id(id_str: str, existing_ids: set) -> str:
    if not id_str.isdigit():
        raise ValueError("ID должен состоять только из цифр.")
    id_int = int(id_str)
    if id_int <= 0:
        raise ValueError("ID должен быть положительным целым числом.")
    formatted_id = f"{id_int:08d}"
    if formatted_id in existing_ids:
        raise ValueError("ID должен быть уникальным.")
    return formatted_id

def print_users_table(users: dict):
    if not users:
        print("Нет данных для отображения.")
        return

    print(f"{'ID':<10} {'Имя':<15} {'Фамилия':<15} {'Возраст':<8}")
    print("-" * 55)
    for user_id, (first_name, last_name, age) in users.items():
        print(f"{user_id:<10} {first_name:<15} {last_name:<15} {age:<8}")

def main():
    users = {}  # Словарь: ID -> (имя, фамилия, возраст)
    existing_ids = set()

    print("Введите данные пользователей. Для завершения оставьте любое поле пустым.")

    while True:
        try:
            # Ввод имени
            first_name_input = input("Имя: ").strip()
            if not first_name_input:
                break

            # Ввод фамилии
            last_name_input = input("Фамилия: ").strip()
            if not last_name_input:
                break

            # Ввод возраста
            age_input = input("Возраст (18–60): ").strip()
            if not age_input:
                break

            # Ввод ID
            id_input = input("ID (целое число): ").strip()
            if not id_input:
                break

            # Обработка и валидация
            first_name = format_name(first_name_input)
            last_name = format_name(last_name_input)
            age = validate_age(age_input)
            user_id = format_id(id_input, existing_ids)

            # Сохранение
            users[user_id] = (first_name, last_name, age)
            existing_ids.add(user_id)

            print("Пользователь добавлен успешно.\n")

        except ValueError as e:
            print(f"Ошибка: {e}\n")
            continue  # Возвращаемся к началу цикла

    # Вывод результата
    print("\nИтоговая таблица пользователей:")
    print_users_table(users)

if __name__ == "__main__":
    main()
```
