# Линейная алгебра, математический анализ

```python
import numpy as np
import matplotlib.pyplot as plt

n = 100
A = np.random.randn(n, n)

U, S, VT = np.linalg.svd(A, full_matrices=False)

errors = []
ranks = range(2, n+1)

for r in ranks:
    # Аппроксимация ранга r
    A_r = U[:, :r] @ np.diag(S[:r]) @ VT[:r, :]
    error = np.linalg.norm(A - A_r, 'fro')
    errors.append(error)

plt.plot(ranks, errors)
plt.xlabel('Ранг r')
plt.ylabel('Ошибка аппроксимации ||A - A_r||_F')
plt.title('Зависимость ошибки от ранга')
plt.grid(True)
plt.show()
```
