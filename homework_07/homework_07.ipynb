### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
# ВАШ КОД ЗДЕСЬ
arrZero = np.zeros(10, dtype=int)
arrZero[4] = 1
print("Одмерный массив")
print(arrZero)
arrTwo = arrZero.reshape(2, 5)
print("Трансформированный массив")
print(arrTwo)
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
# ВАШ КОД ЗДЕСЬ
arrRange = np.arange(10, 50)
arrReverse = arrRange[::-1]
arrEven = arrReverse[arrReverse % 2 == 0]
print(f"Все четные элементы {arrEven}")
print(f"Кол-во четных элементов {len(arrEven)}")
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

```python
# ВАШ КОД ЗДЕСЬ
arrOne = np.arange(9)
arrThree = arrOne.reshape(3,3)
print(f"трех массив {arrThree}")
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
# ВАШ КОД ЗДЕСЬ
np.random.seed(42)
arr_random = np.random.random((4, 3, 2))
print(f"сам массив {arr_random}")
print(f"min - {arr_random.min()} max -  {arr_random.max()}")
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
# ВАШ КОД ЗДЕСЬ
Y = np.random.randint(0, 10, (6, 4))
X = np.random.randint(0, 10, (4, 3))
resultMulti = np.dot(Y,X)
print(f"Результат умножения матриц ")
print(resultMulti)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
# ВАШ КОД ЗДЕСЬ
np.random.seed(42)
arr_two = np.random.rand(7, 7)
meanArr = arr_two.mean()
stdArr = arr_two.std()
print(f"Среднее значение: {meanArr}")
print(f"Стандартное отклонение: {stdArr}")
normalized = (arr_two - meanArr) / stdArr
print(f"нормализованный массив {normalized}")
```

### Найдите количество курящих людей

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
# ВАШ КОД ЗДЕСЬ
import seaborn as sns
tips = sns.load_dataset('tips')
avg_bill_by_day = tips.groupby('day', observed=False)['total_bill'].mean()
print(avg_bill_by_day)
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
# ВАШ КОД ЗДЕСЬ
import seaborn as sns
tips = sns.load_dataset('tips')
median_bill = tips['total_bill'].median()
print(f"Медиана чека: ${median_bill:.2f}\n")
high_bills = tips[tips['total_bill'] > median_bill]
avg_tip_high_bills = high_bills.groupby('sex', observed=False)['tip'].mean()

print("Средние чаевые при чеке выше медианы:")
print(avg_tip_high_bills)
```

### Преобразуйте признак 'smoker' в бинарный (0-No, 1-Yes)

```python
# ВАШ КОД ЗДЕСЬ
import seaborn as sns
tips = sns.load_dataset('tips')
tips['smoker'] = tips['smoker'].map({'Yes': 1, 'No': 0})
print(tips[['smoker']].head())
print(tips['smoker'].dtype)
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
# ВАШ КОД ЗДЕСЬ
import seaborn as sns
import matplotlib.pyplot as plt


tips = sns.load_dataset('tips')
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.histplot(tips['total_bill'], kde=True, color='royalblue', bins=30)

plt.title('Распределение общего счета (Total Bill)')
plt.xlabel('Сумма счета ($)')
plt.ylabel('Количество заказов')

plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
# ВАШ КОД ЗДЕСЬ
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.set_theme(style="darkgrid")
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip', alpha=0.7)

plt.title('Взаимосвязь между общим счетом и чаевыми', fontsize=14)
plt.xlabel('Общий счет ($)')
plt.ylabel('Чаевые ($)')

plt.show()
```

### Постройте pairplot

```python
# ВАШ КОД ЗДЕСЬ
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.set_theme(style="ticks")

sns.pairplot(tips, hue='sex', palette='husl', diag_kind='kde')
plt.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
# ВАШ КОД ЗДЕСЬ
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))
sns.boxplot(data=tips, x='day', y='total_bill', palette='Set3', hue='day', legend=False)

plt.title('Распределение чеков по дням недели', fontsize=15)
plt.xlabel('День недели')
plt.ylabel('Общий счет ($)')
plt.show()
```
