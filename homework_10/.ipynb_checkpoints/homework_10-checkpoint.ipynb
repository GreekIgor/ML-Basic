# Где дешевле жить? Предсказание цен в Airbnb - учимся генерировать признаки и интерпретировать результаты модели

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV, ElasticNetCV
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# 1. ЗАГРУЗКА ДАННЫХ
# Предполагаем, что файл находится в директории homework_10
df = pd.read_csv('AB_NYC_2019.csv')

# --- ЧАСТЬ 1. EDA ---

# Удаление ненужных признаков по условию
drop_cols = ['id', 'name', 'host_id', 'host_name', 'last_review']
df = df.drop(columns=drop_cols)

# Заполнение пропусков в отзывах (если нет отзывов — ставим 0)
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

# Визуализация распределения целевой переменной (Price)
plt.figure(figsize=(10, 5))
sns.histplot(df['price'], bins=100, kde=True)
plt.title('Распределение цен (с сильным смещением)')
plt.show()

# Очистка от экстремальных выбросов (например, цена 0 или выше 1000)
# Это поможет линейным моделям лучше сойтись
df = df[(df['price'] > 0) & (df['price'] < 1000)]

# --- ЧАСТЬ 2. FEATURE ENGINEERING ---

# Создаем признак: Расстояние до Таймс-сквер (центр Манхэттена)
def haversine_distance(lat1, lon1, lat2, lon2):
    # Упрощенная формула Евклидова расстояния для координат
    return np.sqrt((lat1 - lat2)**2 + (lon1 - lon2)**2)

manhattan_lat, manhattan_lon = 40.7580, -73.9855
df['dist_to_center'] = haversine_distance(df['latitude'], df['longitude'], manhattan_lat, manhattan_lon)

# Логарифмируем целевую переменную, чтобы приблизить распределение к нормальному
df['price_log'] = np.log1p(df['price'])

# Определяем признаки
X = df.drop(['price', 'price_log'], axis=1)
y = df['price_log']

# Разделение признаков на категориальные и числовые
cat_features = ['neighbourhood_group', 'neighbourhood', 'room_type']
num_features = ['latitude', 'longitude', 'minimum_nights', 'number_of_reviews', 
                'reviews_per_month', 'calculated_host_listings_count', 
                'availability_365', 'dist_to_center']

# --- ЧАСТЬ 3. ПРЕПРОЦЕССИНГ И МОДЕЛИРОВАНИЕ ---

# Создаем пайплайны для обработки данных
num_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_transformer, num_features),
        ('cat', cat_transformer, cat_features)
    ])

# Разделение на Train/Test (30% на тест)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Список моделей для оценки
models = {
    "Linear Regression": LinearRegression(),
    "RidgeCV": RidgeCV(alphas=np.logspace(-3, 3, 10)),
    "LassoCV": LassoCV(alphas=np.logspace(-3, 3, 10), max_iter=2000),
    "ElasticNetCV": ElasticNetCV(l1_ratio=[.1, .5, .7, .9, .95, .99, 1], max_iter=2000)
}

results = []

for name, model in models.items():
    # Создаем полный пайплайн: Препроцессинг + Модель
    clf = Pipeline(steps=[('preprocessor', preprocessor),
                          ('regressor', model)])
    
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    
    # Возвращаем предсказания из логарифмического масштаба в обычный для метрик MAE/RMSE
    y_test_exp = np.expm1(y_test)
    y_pred_exp = np.expm1(y_pred)
    
    # Считаем метрики
    r2 = r2_score(y_test, y_pred) # R2 лучше считать на логарифмах
    mae = mean_absolute_error(y_test_exp, y_pred_exp)
    rmse = np.sqrt(mean_squared_error(y_test_exp, y_pred_exp))
    
    results.append({"Model": name, "R2": r2, "MAE": mae, "RMSE": rmse})

# Вывод результатов
results_df = pd.DataFrame(results)
print(results_df)

# --- ВИЗУАЛИЗАЦИЯ ВАЖНОСТИ ПРИЗНАКОВ (для RidgeCV) ---
best_model = Pipeline(steps=[('preprocessor', preprocessor),
                             ('regressor', RidgeCV())])
best_model.fit(X_train, y_train)

# Извлекаем названия признаков после OneHotEncoding
ohe_feature_names = best_model.named_steps['preprocessor'].transformers_[1][1]\
    .named_steps['onehot'].get_feature_names_out(cat_features)
all_feature_names = np.concatenate([num_features, ohe_feature_names])

# Берем топ-20 важных коэффициентов
coeffs = best_model.named_steps['regressor'].coef_
importance = pd.Series(coeffs, index=all_feature_names).sort_values(ascending=False)

plt.figure(figsize=(10, 8))
importance.head(10).append(importance.tail(10)).plot(kind='barh')
plt.title('Топ-10 положительных и отрицательных весов признаков (Ridge)')
plt.show()
```
