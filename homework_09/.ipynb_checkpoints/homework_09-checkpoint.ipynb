# Учимся работе с данными - от загрузки до моделирования

```python
#YOUR CODE HERE
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, accuracy_score, precision_score, recall_score, f1_score

```

```python
#Начнем загрузку данных 
# Загружаем данные (предположим, файл лежит в той же папке)
df = pd.read_csv('data.csv')

# Удаляем лишние колонки: 'id' (не несет смысла) и 'Unnamed: 32' (часто появляется в этом датасете как пустая)
df = df.drop(['id', 'Unnamed: 32'], axis=1, errors='ignore')

# Превращаем целевую переменную в числа: M (злокачественная) -> 1, B (доброкачественная) -> 0
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})

print(df.describe().T) # .T транспонирует таблицу для удобства чтения
```

```python
# Возьмем первые 10 признаков для примера тут у нас колонки список со значениями
features_mean = list(df.columns[1:11])

# создать холст определенного размера
plt.figure(figsize=(15, 10))
for i, feature in enumerate(features_mean):
    # создать ячеику где мы поместим свои рисунок график
    plt.subplot(4, 3, i+1)
    sns.histplot(data=df, x=feature, hue='diagnosis', kde=True, palette='viridis')
    # x=feature - столбец который возьмем для горизонтальной оси
    # hue='diagnosis' Этот параметр разделяет данные на группы по значению в столбце diagnosis
    #Вместо одной общей гистограммы Seaborn нарисует две (для M — злокачественных и B — доброкачественных),
    # наложив их друг на друга. Это позволяет мгновенно увидеть, есть ли разница между больными и здоровыми клетками
    # kde=True Рисует плавную линию поверх столбцов гистограммы
    # palette='viridis' Задает цветовую схему. 'viridis' — это профессиональная палитра (от фиолетового до желтого)
    plt.title(f'Распределение {feature}')
    
    # Раздвигает графики друг от друга, если они пересекаются.
    # Подгоняет поля (margins) так, чтобы графики занимали максимум свободного места, но не вылезали за границы холста.
    # Выравнивает подписи, чтобы они были читаемы.
plt.tight_layout()
plt.show()
```

```python
# Кореляция и Тепловая карта
plt.figure(figsize=(12, 10))
# Эта строка превращает скучную таблицу с числами в наглядную тепловую карту. 
# Для датасета Breast Cancer это критически важный шаг, потому что у нас 30 признаков, 
#и многие из них — это просто разные способы измерить одно и то же
# Метод .corr() вычисляет коэффициент корреляции Пирсона между всеми парами столбцов.
# Результат — матрица, где 1.0 означает идеальную прямую зависимость, 0 — отсутствие связи, а -1.0 — обратную зависимость.
# Сокращение от annotation (аннотации). Если поставить True, внутри каждого квадратика будет написано число (значение корреляции)
# Почему мы ставим False? В этом датасете 30 признаков. Если втиснуть 900 чисел ($30 \times 30$) 
#в один график, получится «каша», которую невозможно прочитать
#  Цветовая схема (Colormap)'coolwarm' — идеальный выбор для корреляций:
sns.heatmap(df.corr(), annot=False, cmap='coolwarm')
plt.title("Матрица корреляций")
plt.show()
```

```python
# Линейная зависимость
# Эта команда создает диаграмму рассеяния. Если тепловая карта (Heatmap) сказала нам,
# что между радиусом и периметром есть связь «0.99», то этот график покажет нам эту связь «в лицо».
# x='radius_mean', y='perimeter_mean': Мы берем два признака и расставляем каждую опухоль как точку на плоскости.
# hue='diagnosis': Это самое интересное. Точки окрасятся в два цвета (например, синий и оранжевый).
sns.scatterplot(data=df, x='radius_mean', y='perimeter_mean', hue='diagnosis')
plt.title("Линейная зависимость радиуса и периметра")
plt.show()
```

```python
#Boxplot Боксплоты отлично показывают выбросы и то, насколько "далеко" друг от друга находятся группы.
# По таким признакам как area_worst или concave points_mean разделение очень четкое — это "золотые" признаки для нашей модели.
plt.figure(figsize=(10, 6))
# Если гистограмма показывает нам «горы» данных, то Boxplot (ящик с усами) — это рентген этих гор.
#  Он позволяет мгновенно увидеть медиану, разброс и, что самое важное, выбросы.
# Для задачи классификации рака это, пожалуй, самый полезный график.
# x='diagnosis': Мы делим данные на две колонки: 0 (B — доброкачественные) и 1 (M — злокачественные).
# y='area_worst': Мы смотрим на значение площади самого большого фрагмента опухоли.
sns.boxplot(x='diagnosis', y='area_worst', data=df)
plt.title("Разница area_worst в зависимости от диагноза")
plt.show()
```

```python
# Моделирование (kNN)
# Разбиение на Train/Test Мы разделяем данные, чтобы проверить модель на "незнакомых" ей примерах.
# ы говорим Pandas: «Возьми всю таблицу, но удали из неё колонку с названием diagnosis»
X = df.drop('diagnosis', axis=1) # Признаки
y = df['diagnosis']              # Цель

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Зачем это нужно для kNN?
# Алгоритм kNN считает расстояние (евклидово) между точками. Если один признак измеряется в тысячах (площадь),
# а другой — в сотых долях (гладкость), то площадь "задавит" гладкость своим масштабом.
# Стандартизация приводит все к одному масштабу (среднее 0, отклонение 1).

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test) # Используем параметры обучения для теста!
```

```python
#Модель "из коробки"
# Обучаем kNN с параметрами по умолчанию (обычно 5 соседей).
knn_base = KNeighborsClassifier()
knn_base.fit(X_train_scaled, y_train)
y_pred = knn_base.predict(X_test_scaled)

# Метрики
print(f"Accuracy: {accuracy_score(y_test, y_pred):.3f}")
print(classification_report(y_test, y_pred))
```

```python
# ROC-кривая
# ROC-кривая показывает баланс между чувствительностью (Recall) и специфичностью.
# Чем больше площадь под кривой (AUC), тем лучше модель.

y_score = knn_base.predict_proba(X_test_scaled)[:,1]
fpr, tpr, _ = roc_curve(y_test, y_score)
roc_auc = auc(fpr, tpr)

plt.plot(fpr, tpr, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.legend()
plt.show()
```

```python
# Настройка гиперпараметра (Grid Search)Попробуем найти идеальное количество соседей $k$.
# Ищем k в диапазоне от 1 до 30
param_grid = {'n_neighbors': np.arange(1, 31)}
grid_search = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5)
grid_search.fit(X_train_scaled, y_train)

print(f"Лучшее число соседей: {grid_search.best_params_}")

# Оценка финальной модели
final_model = grid_search.best_estimator_
y_pred_final = final_model.predict(X_test_scaled)
print(f"Новый Accuracy: {accuracy_score(y_test, y_pred_final):.3f}")
```

```python
# Итоги:
# EDA помог нам понять, что признаки размера и вогнутости (concavity) — самые важные.
# Стандартизация критична для kNN, без нее точность была бы гораздо ниже.
# Кросс-валидация позволила нам не "гадать" количество соседей, а выбрать оптимальное на основе данных.
# Как думаешь, какой из признаков оказался самым "говорящим" в твоем анализе?

```
